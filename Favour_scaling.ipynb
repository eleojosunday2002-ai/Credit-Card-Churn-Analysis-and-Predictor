Inspecting the Dataset Before Scaling

Before applying the scaling procedure, the structure and dimensions of the dataset are examined.

Feature Scaling:

We’re going to use the StandardScaler because it uses standardization (with each feature centered around a mean of approximately 0 and scaled to a standard deviation of approximately 1) which is particularly appropriate when features have different units and magnitudes and when we want them to contribute on a comparable standardized scale. This is important because K-Means uses distances to form clusters and PCA is sensitive to the scale of the input variables. Without scaling, variables with larger numerical magnitudes, such as transaction amounts and credit limits, could dominate variables measured on smaller scales.

```python
df.head()
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

X_scaled = scaler.fit_transform(df)
```

```python
scaled_df = pd.DataFrame(
    X_scaled,
    columns=df.columns,
    index=df.index)
```

Confirming the scaled data.

```python
scaled_df.head()
```

```python
scaled_df.to_csv(
    "scaled_data_for_clustering.csv",
    index=False)
```
